## 🎯 What This Module Does

This notebook demonstrates how the API's configuration is **centralized** using **Pydantic Settings** — a best practice from the **12-Factor App** methodology. All configurable values (MLflow URI, model name, feature paths, server settings) live in one place, making the service **portable, testable, and production-ready**.

## 🏗️ Why Centralized Configuration Matters

| Problem | Solution |
|---------|----------|
| Hardcoded paths break when the project moves | `Path(__file__)` resolves paths dynamically |
| Model names scattered across files | Single `MODEL_NAME` setting |
| Environment-specific values (dev/prod) | `.env` file support via Pydantic |
| No validation of config values | Pydantic validates types & ranges at import |

---

## 📋 Key Configuration Settings

| Setting | Value | Purpose |
|---------|-------|---------|
| `MLFLOW_TRACKING_URI` | `sqlite:///.../notebooks/mlflow.db` | Points to the MLflow database with registered models |
| `MODEL_NAME` | `Random Forest` | The registered model served by default |
| `MODEL_STAGE` | `Production` | Which stage to load the model from |
| `FEATURE_COLUMNS_PATH` | `.../processed_data/feature_columns.json` | 35 feature names in training order |
| `PREDICTION_THRESHOLD` | `0.5` | Confidence threshold for positive classification |
| `HOST` | `0.0.0.0` | Bind to all network interfaces |
| `PORT` | `8000` | FastAPI server port |

## 🧠 Skills Demonstrated

- ✅ **Pydantic Settings** — type-safe configuration
- ✅ **12-Factor App** — config in environment, not code
- ✅ **Path Resolution** — portable, machine-independent paths
- ✅ **Singleton Pattern** — one shared settings instance

```python
# ── Add the api directory to sys.path so 'app' module can be found ──
import sys
from pathlib import Path

api_dir = Path.cwd() / "cervical-cancer-mlops" / "api"
if str(api_dir) not in sys.path:
    sys.path.insert(0, str(api_dir))
print(f"✅ Added {api_dir} to sys.path")
```

```python
# ── Import the centralized settings ──
from api_config.py import settings

print("=" * 60)
print("  RESOLVED CONFIGURATION VALUES")
print("=" * 60)
print(f"MLflow Tracking URI:  {settings.MLFLOW_TRACKING_URI}")
print(f"Model Name:           {settings.MODEL_NAME}")
print(f"Model Stage:          {settings.MODEL_STAGE}")
print(f"Feature Columns Path: {settings.FEATURE_COLUMNS_PATH}")
print(f"Prediction Threshold: {settings.PREDICTION_THRESHOLD}")
print(f"Server Host:          {settings.HOST}")
print(f"Server Port:          {settings.PORT}")
print(f"Auto-Reload:          {settings.RELOAD}")
```

```python
# ── Verify the feature columns file exists and count features ──
import json
from pathlib import Path

feature_path = Path(settings.FEATURE_COLUMNS_PATH)
if feature_path.exists():
    with open(feature_path, "r") as f:
        feature_columns = json.load(f)
    print(f"✅ Feature columns file found: {feature_path}")
    print(f"📊 Total features: {len(feature_columns)}")
    print(f"\nFirst 5 features: {feature_columns[:5]}")
    print(f"Last 5 features:  {feature_columns[-5:]}")
else:
    print(f"❌ Feature columns file NOT found: {feature_path}")
```

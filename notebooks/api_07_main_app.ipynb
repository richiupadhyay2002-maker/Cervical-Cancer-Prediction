## 🎯 What This Module Does

This notebook demonstrates how the **FastAPI application is assembled** — the central factory that wires together all the components (routers, middleware, exception handlers, and the model-loading lifespan).

## 🏗️ Application Startup Flow

```
┌─────────────────────────────────────────────────────────────┐
│                    APP STARTUP (lifespan)                  │
├─────────────────────────────────────────────────────────────┤
│  1. Log startup banner                                      │
│  2. Load default model from MLflow Registry                 │
│  3. If model fails → log error, server still starts         │
│  4. Register CORS middleware                                │
│  5. Register routers (health, predict, report)              │
│  6. Register exception handlers (422, 500)                  │
│  7. Ready to serve on http://0.0.0.0:8000                  │
└─────────────────────────────────────────────────────────────┘
```

## 💡 Why Each Component Matters

| Component | Purpose |
|-----------|---------|
| **Lifespan** | Loads model once at startup, not per-request |
| **CORS Middleware** | Allows web frontends to call the API |
| **Routers** | Modular endpoint organization |
| **422 Handler** | Clean validation error responses |
| **500 Handler** | Sanitized server error responses |
| **Root redirect** | `/` → `/docs` for easy access |

````python
# ── Add the api directory to sys.path so 'app' module can be found ──
import sys
from pathlib import Path

_candidates = [
    Path.cwd() / "cervical-cancer-mlops" / "api",
    Path.cwd().parent / "api",
    Path.cwd() / "api",
]
api_dir = next((p for p in _candidates if (p / "app").exists()), None)
if api_dir is None:
    raise RuntimeError(
        "Could not locate the 'api/app' package. "
        "Expected it under one of: " + ", ".join(map(str, _candidates))
    )
if str(api_dir) not in sys.path:
    sys.path.insert(0, str(api_dir))
print(f"✅ Added {api_dir} to sys.path")
````

````python
# ── Import the FastAPI application ──
from app.main import app

print("✅ FastAPI application imported!")
print(f"   Title:       {app.title}")
print(f"   Version:     {app.version}")
print(f"   Docs URL:    {app.docs_url}")
print(f"   Redoc URL:   {app.redoc_url}")
````

````python
# ── Show all registered routes ──
print("📋 ALL REGISTERED ROUTES")
print("=" * 60)
for route in app.routes:
    if hasattr(route, "methods"):
        methods = ",".join(sorted(route.methods))
        print(f"  {methods:<20} {route.path}")
    else:
        print(f"  {'MOUNT':<20} {route.path}")
````

````python
# ── Show middleware stack ──
print("🛡️ MIDDLEWARE STACK")
print("=" * 60)
for middleware in app.user_middleware:
    print(f"  - {middleware.cls.__name__}")

print("\n✅ CORS configured:")
print(f"   allow_origins:  {app.middleware_stack}")
print(f"   allow_methods:  *")
````

````python
# ── Show exception handlers ──
print("🚨 EXCEPTION HANDLERS")
print("=" * 60)
for exc_type, handler in app.exception_handlers.items():
    if exc_type is not None:
        print(f"  - {exc_type.__name__:<30} → {handler.__name__}")

print("\n✅ Handlers registered:")
print("   - 422 ValidationError → clean JSON response")
print("   - 500 Exception → sanitized JSON response")
````

## 🎯 What This Module Does

This notebook demonstrates how the API **loads machine learning models from the MLflow Model Registry** — a production-grade model management system. Models are loaded on-demand, cached in memory for performance, and can be switched dynamically via query parameters.

## 🏗️ Architecture

```
┌─────────────┐     ┌──────────────────┐     ┌─────────────────┐
│  FastAPI    │────▶│  Model Loader    │────▶│  MLflow Registry │
│  Endpoint   │     │  (cached)        │     │  (SQLite DB)     │
└─────────────┘     └──────────────────┘     └─────────────────┘
       │                     │                        │
       │                     ▼                        ▼
       │              ┌─────────────┐          ┌─────────────┐
       └──────────────│  Prediction │          │  8 Models   │
                      │  Response   │          │  Registered │
                      └─────────────┘          └─────────────┘
```

## 💡 Why Model Registry Matters

| Feature | Benefit |
|---------|---------|
| **Versioning** | Every model has a version — rollback is trivial |
| **Staging** | Models move through None → Staging → Production |
| **Caching** | Models loaded once, reused for all requests |
| **Dynamic switching** | `?model=AdaBoost` switches models at runtime |
| **Fallback** | Loads latest version if stage not found |

````python
# ── Add the api directory to sys.path so 'app' module can be found ──
import sys
from pathlib import Path

_candidates = [
    Path.cwd() / "cervical-cancer-mlops" / "api",
    Path.cwd().parent / "api",
    Path.cwd() / "api",
]
api_dir = next((p for p in _candidates if (p / "app").exists()), None)
if api_dir is None:
    raise RuntimeError(
        "Could not locate the 'api/app' package. "
        "Expected it under one of: " + ", ".join(map(str, _candidates))
    )
if str(api_dir) not in sys.path:
    sys.path.insert(0, str(api_dir))
print(f"✅ Added {api_dir} to sys.path")
````

````python
# ── Import the model loader ──
from app.model_loader import (
    load_default_model,
    get_model,
    get_default_model,
    list_available_models,
    get_native_model,
    has_probability_support,
    get_prediction_confidence,
)

print("✅ Imported all model loader functions")
````

````python
# ── List all registered models in the MLflow Registry ──
models = list_available_models()

print(f"📊 Found {len(models)} registered models:\n")
print(f"{'Model Name':<25} {'Version':<8} {'Stage':<12}")
print("-" * 50)
for m in models:
    print(f"{m['name']:<25} v{m['version']:<7} {m['stage']:<12}")
````

````python
# ── Load the default model (Random Forest) ──
load_default_model()
model, version, stage, name = get_default_model()

print(f"✅ Default model loaded:")
print(f"   Name:    {name}")
print(f"   Version: {version}")
print(f"   Stage:   {stage}")
print(f"   Type:    {type(model).__name__}")

# Check probability support
print(f"\n🔍 Probability support: {has_probability_support(model)}")
````

````python
# ── Demonstrate dynamic model switching ──
# Load a different model on-demand (AdaBoost is in 'None' stage)
ada_model, ada_version, ada_stage, ada_name = get_model("AdaBoost")

print(f"✅ Dynamically loaded '{ada_name}':")
print(f"   Version: {ada_version}")
print(f"   Stage:   {ada_stage}")
print(f"   Type:    {type(ada_model).__name__}")
print(f"   Probability support: {has_probability_support(ada_model)}")
````
